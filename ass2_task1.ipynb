```python
# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
```

```python
## data cleaning

df = pd.read_csv('./CAB330_asm2_datasets/CAB330_asm2_datasets/Dataset4Task1_26.csv', na_filter=False)
df.info()


# check for duplicates
print(df[df.duplicated(keep=False)])
print("#### dupe check ####")
dupe_count = df.duplicated().sum()
print(dupe_count)
initial_shape = df.shape[0]
# export to observe duplicate entries
duplicates = df[df.duplicated(keep=False)]
duplicates = duplicates.sort_values(by=list(df.columns))
duplicates.to_excel('dupes.xlsx')


# convert data type from object to float. 
# conversion failed because of empty string entries
# scan to see number of empty strings

print("\n### Search for empty string entry ###")
count = 0
total_count = 0

for col in df.columns:
    count = (df[col] == '').sum()
    total_count += count
    if (count != 0):
        print(f"*{col}* empty count: {count}")
print(f"total empty count: {total_count}")

# convert empty entry into NaN

df = df.replace('', np.nan)

print("\n#### Check for NaN values ####")
null_count = 0
for col in df.columns:
    temp_count = df[col].isna().sum()
    null_count += temp_count

print(null_count)


# application of data cleaning process
df = df.drop_duplicates()
print("\n#### Duplication removal check ####")
dupe_count = df.duplicated().sum()
print(dupe_count)  # confirmed to have zero duplication

# change data type to float
df = df.astype(float)
df.info()

# addressing NaN values
to_be_deleted = []
for index, row in df.iterrows():
    row_null = 0
    for column in df.columns:
        if (pd.isna(row[column])):
            row_null += 1
    if row_null >= 2:
        to_be_deleted.append(index)
print(to_be_deleted)

# graph plot for each data point
for col in df.columns:
    sns.distplot(df[col].dropna())
    plt.title(f"Distribution of {col}")
    plt.show()

# matrix of graphs
cols = ['fixed acidity', 'citric acid', 'total sulfur dioxide', 'alcohol']
fig, axes = plt.subplots(2, 2, figsize=(10, 10))

for axe, col in zip(axes.flat, cols):
    sns.histplot(df[col], kde=True, ax=axe)
    axe.set_title(f"Distribution of {col}")
plt.tight_layout()
plt.savefig('matrix.png', dpi=300, bbox_inches='tight')
plt.show()

#fill with median
df = df.fillna(df.median())

df.info()
final_shape = df.shape[0]

print("### before and after shape ###")
print(initial_shape)
print(final_shape)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings("ignore")

scaler = StandardScaler()
X = scaler.fit_transform(df)

rs = 42
# 1.2: all variable K-means 
k_range = range(2, 17)

# stores cluster index, inertia and Silhouette score in that order 
clusters={}

for index in k_range:
    model = KMeans(n_clusters=index, random_state=rs)
    model.fit(X)

    inertia = model.inertia_
    sil_score = silhouette_score(X, model.labels_)

    clusters[index] = [inertia, sil_score]

for cluster, (inertia, sil_score) in clusters.items():
    print(f"for cluster {cluster}: inertia {inertia} & sil score {sil_score}")

# extract into seperate lists
cluster_list = list(clusters.keys())
inertia_list = []
sil_list = []
for i in clusters:
    inertia_list.append(clusters[i][0])
    sil_list.append(clusters[i][1])



plt.figure(figsize = (6,5))
plt.title("Elbow method graph")
plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.plot(cluster_list, inertia_list, marker="*")

plt.figure(figsize = (6,5))
plt.title("silhouette_score graph")
plt.xlabel("Number of clusters")
plt.ylabel("silhouette_score")
plt.plot(cluster_list, sil_list, marker="*")

# k = 4 was selected
#tabulation
results = pd.DataFrame({"cluster #": list(k_range), "Inertia": inertia_list, "Silhouette": sil_list})
print(results.round({"Inertia": 2, "Silhouette": 4}).to_string(index=False))
```

```python
# part 3: 
matrix = df.corr()
print(matrix)
matrix.to_excel('corr.xlsx')
# free sulfur dioxide - explainable through tutal sulfur dioxide so redundant
# chlorides - quality indicator, a seperate measure from profiling such as 
# # density - explainable though other features such as sugar and alcohol 
# ph - derived from acidity measurements

plt.figure(figsize=(6,6))
sns.heatmap(matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')

df2 = df[['fixed acidity', 'volatile acidity', 'citric acid', 'residual sugar', 'total sulfur dioxide', 'sulphates', 'alcohol']] 

# re-fit and display
k_range = range(2, 17)
scaler2 = StandardScaler()
X_2 = scaler2.fit_transform(df2)
sil2_list = []
in2_list = []

for k in k_range:
    model = KMeans(n_clusters=k, n_init = 12, random_state=rs)
    labels = model.fit_predict(X_2)
    in2_list.append(model.inertia_)
    sil2_list.append(silhouette_score(X_2, labels))

plt.savefig('corr_heatmap.png', dpi=300, bbox_inches='tight')

#tabulation
results = pd.DataFrame({"cluster #": list(k_range), "Inertia": in2_list, "Silhouette": sil2_list})
print(results.round({"Inertia": 2, "Silhouette": 5}).to_string(index=False))
```
